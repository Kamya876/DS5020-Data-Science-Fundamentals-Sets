```python
import pandas as pd


df = pd.read_csv("titanic-1.csv")


U  = set(df["PassengerId"])
F  = set(df[df["Sex"] == "female"]["PassengerId"])
M  = set(df[df["Sex"] == "male"]["PassengerId"])
C1 = set(df[df["Pclass"] == 1]["PassengerId"])
C2 = set(df[df["Pclass"] == 2]["PassengerId"])
C3 = set(df[df["Pclass"] == 3]["PassengerId"])
S  = set(df[df["Survived"] == 1]["PassengerId"])
K  = set(df[df["Age"] < 18]["PassengerId"])

print("|U|  =", len(U))
print("|F|  =", len(F),  "|M|  =", len(M))
print("|C1| =", len(C1), "|C2| =", len(C2), "|C3| =", len(C3))
print("|S|  =", len(S),  "|K|  =", len(K))
```

## Part A: The Empty Set and the Universal Set

**A1.** How many passengers are in the universal set U?

**A2.** Let A = {passengers older than 100}. Is A the empty set?

**A3.** Is F ∪ M equal to U?

```python
A = set(df[df["Age"] > 100]["PassengerId"])

print("A1 |U| =", len(U))
print("A2 oldest age =", df["Age"].max(), "| A empty?", len(A) == 0)
print("A3 F ∪ M == U?", (F | M) == U)
```

**Answers:**
* **A1:** There are **891** passengers in the universal set.
* **A2:** **Yes**, A is the empty set. The oldest passenger in the dataset is 80 years old, so no one is older than 100. 
* **A3:** **Yes**, F ∪ M equals U. Every passenger is either female (314) or male (577), which adds up to exactly 891.

```python
print("B1 |F ∪ C1| =", len(F | C1))
print("B2 |S ∪ K|  =", len(S | K))
print("B3 check:", len(F) + len(C1) - len(F & C1), "== len(F ∪ C1)?", 
      len(F) + len(C1) - len(F & C1) == len(F | C1))
```

**Answers:**
* **B1:** |F ∪ C1| = **436** passengers.
* **B2:** |S ∪ K| = **394** passengers.
* **B3:** The check is True. 314 (Female) + 216 (1st Class) − 94 (Both) = **436**.

## Part C: Intersection

**C1.** How many passengers are female AND in 1st class?

**C2.** How many female passengers survived?

**C3.** How many 3rd-class male passengers survived?

```python
print("C1 |F ∩ C1|      =", len(F & C1))
print("C2 |F ∩ S|       =", len(F & S))
print("C3 |C3 ∩ M ∩ S|  =", len(C3 & M & S))
```

**Answers:**

## Part D: Complement, Difference, and Cartesian Product

**D1.** How many passengers did NOT survive? (S′)

**D2.** How many 1st-class passengers did NOT survive? (C1 − S)

**D3.** Is C1 − S the same as S − C1? Find both.

**D4.** Let X = {female, male} and Y = {1, 2, 3}. List X × Y and find its size.

```python
print("D1 |S'|     =", len(U - S))
print("D2 |C1 - S| =", len(C1 - S))
print("D3 C1-S =", len(C1 - S), "| S-C1 =", len(S - C1), "| same?", (C1 - S) == (S - C1))

X = {"female", "male"}
Y = {1, 2, 3}
XY = {(x, y) for x in X for y in Y}
print("D4 X × Y =", XY)
print("D4 |X × Y| =", len(XY))
```

## Part E: Disjoint Sets and Partitions

**E1.** Are F and M disjoint?

**E2.** Are F and C1 disjoint?

**E3.** Do {C1, C2, C3} form a partition of U?

**E4.** Do the ports of embarkation {C, Q, S} form a partition of U? Explain.

```python
print("E1 F ∩ M empty?", (F & M) == set())
print("E2 F ∩ C1 empty?", (F & C1) == set())
print("E3 C1∪C2∪C3 == U?", (C1 | C2 | C3) == U, "| sum =", len(C1) + len(C2) + len(C3))

ports = set(df["Embarked"].dropna())
print("E4 ports =", ports, "| total in ports =", len(df[df["Embarked"].isin(ports)]), "| missing =", df["Embarked"].isna().sum())
print("    Embarked counts:", df["Embarked"].value_counts(dropna=False).to_dict())
```

**Answers:**
* **E1:** **Yes**, they are disjoint. No passenger can be both female and male. (F ∩ M = ∅)
* **E2:** **No**, they are not disjoint. 94 passengers are both female and in 1st class.
* **E3:** **Yes**, they form a partition. They do not overlap (a passenger is only in one class), and together they add up to exactly 891 (the whole set U).
* **E4:** **No**, they do not form a partition. There are **2 passengers** missing their `Embarked` data, so the ports {C, Q, S} do not cover the entire universal set U.

## Part F: Laws of Set Operations

**F1.** Commutative law: Is F ∩ S = S ∩ F?
**F2.** De Morgan’s law: Is (F ∪ C1)′ = F′ ∩ C1′? How many passengers is that?
**F3.** Distributive law: Is S ∩ (F ∪ C1) = (S ∩ F) ∪ (S ∩ C1)?

```python
print("F1 F ∩ S == S ∩ F?", (F & S) == (S & F), "| size =", len(F & S))

left2  = U - (F | C1)
right2 = (U - F) & (U - C1)
print("F2 (F ∪ C1)' size =", len(left2), "| F' ∩ C1' size =", len(right2), "| equal?", left2 == right2)

left3  = S & (F | C1)
right3 = (S & F) | (S & C1)
print("F3 left =", len(left3), "| right =", len(right3), "| equal?", left3 == right3)
```

**Answers:**
* **F1:** **Yes**, the commutative law holds. F ∩ S equals S ∩ F (both result in exactly 233 passengers).
* **F2:** **Yes**, De Morgan's law holds. Both (F ∪ C1)′ and F′ ∩ C1′ result in exactly **455** passengers (the complement of the 436 passengers in F ∪ C1).
* **F3:** **Yes**, the distributive law holds. Both sides equal exactly **278** passengers, proving the law works on this dataset.
